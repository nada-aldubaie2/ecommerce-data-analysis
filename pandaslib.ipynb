```python
import pandas as pd

# 1. تعريف المستويات: الدولة والسنة
countries = ["مصر", "السعودية"]
years = [2022, 2023]

# استخدام pd.MultiIndex.from_product لدمج كل الدول مع كل السنوات تلقائياً
index = pd.MultiIndex.from_product(
    [countries, years], names=["Country", "Year"]
)

# 2. البيانات المقابلة لهذه الفهارس (4 صفوف إجمالاً)
data = {"Sales": [15000, 22000, 30000, 38000], "Profit": [3000, 5000, 7000, 9500]}

# 3. إنشاء الـ DataFrame
df = pd.DataFrame(data, index=index)

print("--- الجدول الأصلي بـ MultiIndex ---")
print(df)
```
